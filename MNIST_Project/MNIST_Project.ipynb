 ## Step 1:Setup & Baseline Models
- [✅] Set up shared Google Colab notebook
- [✅] Load MNIST dataset (shape 28x28)
- [✅] Shuffle dataset
- [✅] Implement basic ANN
- [✅] Implement SVM
- [✅] Document accuracy, layers, params, train/test time for both

```python
import time
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.svm import SVC
from tensorflow import keras
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.models import Sequential

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize pixel values
x_train = x_train.astype("float32") / 255
x_test = x_test.astype("float32") / 255

# Shuffle training data properly
shuffle_index = np.random.permutation(len(x_train))
x_train = x_train[shuffle_index]
y_train = y_train[shuffle_index]

# Prepare flattened data for potential SVM use (optional)
x_train_svm = x_train.reshape(-1, 28 * 28)
x_test_svm = x_test.reshape(-1, 28 * 28)

# --- ANN Implementation ---
print("Starting ANN training...")
start_time_ann = time.time()

model_ann = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model_ann.compile(optimizer='adam',
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

model_ann.fit(x_train, y_train, epochs=5, batch_size=32, verbose=0)

end_time_ann = time.time()
train_time_ann = end_time_ann - start_time_ann

y_pred_ann = np.argmax(model_ann.predict(x_test, verbose=0), axis=1)
accuracy_ann = accuracy_score(y_test, y_pred_ann)

print("ANN - Accuracy:", accuracy_ann)
print("ANN - Layers:", len(model_ann.layers))
print("ANN - Trainable Parameters:", model_ann.count_params())
print("ANN - Training Time:", train_time_ann, "seconds")

#--- SVM Implementation ---
print("\nStarting SVM training...")
start_time_svm = time.time()
model_svm = SVC(kernel='linear', C=1) # Reduced C for demonstration

# Fit the model
model_svm.fit(x_train_svm, y_train)

end_time_svm = time.time()
train_time_svm = end_time_svm - start_time_svm

y_pred_svm = model_svm.predict(x_test_svm)
accuracy_svm = accuracy_score(y_test, y_pred_svm)


print("SVM - Accuracy:", accuracy_svm)
print("SVM - Kernel:", model_svm.kernel)
print("SVM - Training Time:", train_time_svm, "seconds")
```
